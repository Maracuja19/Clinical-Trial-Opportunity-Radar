Which therapeutic areas and country–indication combinations represent the most attractive opportunities for clinical trial development, considering unmet opportunity, competition and feasibility?

```python
import pandas as pd 
import numpy as np
from itertools import product 
```

```python
#countries
countries = [
    "Argentina",
    "Brazil",
    "Chile",
    "Colombia",
    "Ecuador",
    "Mexico",
    "Paraguay",
    "Uruguay",
    "Venezuela"
]
```

```python
#indication/area
indications_by_area = {
    "Neurology": [
        "Multiple sclerosis",
        "Parkinson's disease",
        "Amyotrophic lateral sclerosis"
    ],
    "Rheumatology": [
        "Rheumatoid arthritis",
        "Systemic lupus erythematosus",
        "Axial spondyloarthritis"
    ],
    "Oncology": [
        "Non-small cell lung cancer",
        "Breast cancer",
        "Colorectal cancer"
    ],
    "Respiratory": [
        "Asthma",
        "Chronic obstructive pulmonary disease",
        "Idiopathic pulmonary fibrosis"
    ],
    "Dermatology": [
        "Atopic dermatitis",
        "Plaque psoriasis",
        "Acne vulgaris"
    ]
}
```

```python
#indication list - rel bet TA and indication
indications = [
    (therap_area, indication)
    for therap_area, indication_list in indications_by_area.items()
    for indication in indication_list
]
```

```python
combinations = [
    (country, therap_area, indication)
    for country, (therap_area, indication) in product(countries, indications)
]
```

```python
#dataframe
df = pd.DataFrame(
    combinations,
    columns= ["country", "therap_area", "indication"]
)

print(f"N° of rows: {len(df)}")
print(f"N° of countries: {df['country'].nunique()}")
print(f"N° of indications: {df['indication'].nunique()}")

display(df.head(10))

print(f"Duplicate country-indication pairs: {df.duplicated(subset=['country', 'indication']).sum()}")
print(f"Missing values: {df.isna().sum().sum()}")

display(df.groupby("therap_area")["indication"].nunique())
```

```python
rng = np.random.default_rng(42)
```

```python
#synthetic parameters
country_profiles = {
    "Argentina":  {"population_factor": 0.55, "research_intensity": 0.90, "reg_baseline": 70, "feas_baseline": 72},
    "Brazil":     {"population_factor": 1.00, "research_intensity": 1.00, "reg_baseline": 72, "feas_baseline": 75},
    "Chile":      {"population_factor": 0.30, "research_intensity": 0.65, "reg_baseline": 75, "feas_baseline": 76},
    "Colombia":   {"population_factor": 0.50, "research_intensity": 0.70, "reg_baseline": 65, "feas_baseline": 68},
    "Ecuador":    {"population_factor": 0.25, "research_intensity": 0.45, "reg_baseline": 62, "feas_baseline": 64},
    "Mexico":     {"population_factor": 0.80, "research_intensity": 0.85, "reg_baseline": 68, "feas_baseline": 70},
    "Paraguay":   {"population_factor": 0.15, "research_intensity": 0.30, "reg_baseline": 58, "feas_baseline": 60},
    "Uruguay":    {"population_factor": 0.10, "research_intensity": 0.40, "reg_baseline": 74, "feas_baseline": 73},
    "Venezuela":  {"population_factor": 0.40, "research_intensity": 0.35, "reg_baseline": 50, "feas_baseline": 52}
}
```

```python
indication_profiles = {
    "Multiple sclerosis": {"population_factor": 0.25, "unmet_need": 70, "competition": 0.60, "complexity": 0.65},
    "Parkinson's disease": {"population_factor": 0.35, "unmet_need": 75, "competition": 0.65, "complexity": 0.60},
    "Amyotrophic lateral sclerosis": {"population_factor": 0.08, "unmet_need": 95, "competition": 0.30, "complexity": 0.90},
    "Rheumatoid arthritis": {"population_factor": 0.50, "unmet_need": 60, "competition": 0.85, "complexity": 0.45},
    "Systemic lupus erythematosus": {"population_factor": 0.30, "unmet_need": 85, "competition": 0.55, "complexity": 0.75},
    "Axial spondyloarthritis": {"population_factor": 0.35, "unmet_need": 65, "competition": 0.45, "complexity": 0.50},
    "Non-small cell lung cancer": {"population_factor": 0.20, "unmet_need": 85, "competition": 0.95, "complexity": 0.85},
    "Breast cancer": {"population_factor": 0.45, "unmet_need": 65, "competition": 1.00, "complexity": 0.65},
    "Colorectal cancer": {"population_factor": 0.35, "unmet_need": 70, "competition": 0.85, "complexity": 0.70},
    "Asthma": {"population_factor": 0.90, "unmet_need": 45, "competition": 0.90, "complexity": 0.35},
    "Chronic obstructive pulmonary disease": {"population_factor": 0.70, "unmet_need": 65, "competition": 0.70, "complexity": 0.55},
    "Idiopathic pulmonary fibrosis": {"population_factor": 0.10, "unmet_need": 90, "competition": 0.35, "complexity": 0.85},
    "Atopic dermatitis": {"population_factor": 0.80, "unmet_need": 55, "competition": 0.80, "complexity": 0.35},
    "Plaque psoriasis": {"population_factor": 0.60, "unmet_need": 55, "competition": 0.85, "complexity": 0.40},
    "Acne vulgaris": {"population_factor": 1.00, "unmet_need": 35, "competition": 0.90, "complexity": 0.25}
}
```

```python
#patient pop creation (thousand peopl), combination from country factor with indication factor and random variation to each
df["patient_pop"] = df.apply(
    lambda row: max(
        1000,
        min(
            100000,
            100000
            * country_profiles[row["country"]]["population_factor"]
            * indication_profiles[row["indication"]]["population_factor"]
            * rng.normal(1.0, 0.08)
        )
    ),
    axis=1
)

df["patient_pop"] = df["patient_pop"].round(0).astype(int)

display(df[["country", "indication", "patient_pop"]].head(10))
```

```python
#med need  per indication (%)
df["unmet_need"] = df.apply(
    lambda row: np.clip(
        indication_profiles[row["indication"]]["unmet_need"]
        + rng.normal(0, 4),
        1,
        100
    ),
    axis=1
)

df["unmet_need"] = df["unmet_need"].round(0).astype(int)

display(df[["country", "indication", "unmet_need"]].head(10))
```

```python
#competition -> it depends on the relative research intensity of each country and the intensity associated to the indication 
df["trial_compet"] = df.apply(
    lambda row: min(
        7,
        max(
            0,
            int(round(
                10
                * country_profiles[row["country"]]["research_intensity"]
                * indication_profiles[row["indication"]]["competition"]
                * rng.normal(1.0, 0.15)
            ))
        )
    ),
    axis=1
)

display(df[["country", "indication", "trial_compet"]].head(10))
```

```python
#feas
df["op_feas"] = df.apply(
    lambda row: np.clip(
        country_profiles[row["country"]]["feas_baseline"]
        - 20 * indication_profiles[row["indication"]]["complexity"]
        + rng.normal(0, 4),
        1,
        100
    ),
    axis=1
)

df["op_feas"] = df["op_feas"].round(0).astype(int)

display(df[["country", "indication", "op_feas"]].head(10))
```

```python
#reg environment
df["reg_env"] = df["country"].map(
    lambda country: country_profiles[country]["reg_baseline"]
)

display(
    df[["country", "reg_env"]]
    .drop_duplicates(subset="country")
    .sort_values("country")
)
```

```python
reg_variation = df.groupby("country")["reg_env"].nunique()

print("Countries with more than one regulatory score:",
      (reg_variation > 1).sum())

display(
    df[["country", "reg_env"]]
    .drop_duplicates()
    .sort_values("country")
    .reset_index(drop=True)
)
```
